```python
import json
from googletrans import Translator
import eng_to_ipa as p
from wordfreq import top_n_list

# 1. Lấy danh sách từ vựng THÔNG DỤNG NHẤT (Top 15.000 từ phổ biến nhất)
# Điều này đảm bảo không bao giờ bị dính các từ vô nghĩa, từ cổ.
common_words = top_n_list('en', 15000)

# 2. Đọc file cũ của bạn
file_name = 'E:/test/speak joy/be/get_data/kids_english_dataset_5000_FINAL.json' # Thay bằng tên file hiện tại của bạn nếu khác
try:
    with open(file_name, 'r', encoding='utf-8') as f:
        current_data = json.load(f)
    print(f"Đã tải thành công {len(current_data)} từ từ file cũ.")
except FileNotFoundError:
    print(f"Không tìm thấy file {file_name}. Hệ thống sẽ tạo từ đầu.")
    current_data = []

# Lấy danh sách các từ đã có để tránh trùng lặp
existing_words = set(item['word'].lower() for item in current_data)

# 3. Lọc các từ mới phù hợp cho trẻ em (từ 3 đến 8 chữ cái, và phải là từ mới)
# Chỉ lấy các từ chỉ chứa chữ cái alphabe (loại bỏ các từ có chứa số hay dấu nháy đơn)
kid_words = [
    w.lower() for w in common_words 
    if 3 <= len(w) <= 8 
    and w.lower() not in existing_words
    and w.isalpha()
]

translator = Translator()

start_id = current_data[-1]['id'] + 1 if current_data else 1
words_needed = 5000 # Số lượng từ bạn muốn tạo thêm

# Đảm bảo chúng ta có đủ từ để tạo
if len(kid_words) < words_needed:
    print(f"Chỉ tìm thấy {len(kid_words)} từ thông dụng mới. Sẽ tạo ngần này từ.")
    words_needed = len(kid_words)

new_entries = []
print(f"Đang tự động tạo thêm {words_needed} từ vựng mới (ID từ {start_id} đến {start_id + words_needed - 1})...")

selected_words = kid_words[:words_needed]

# 4. Bắt đầu vòng lặp tạo từ
for i, word in enumerate(selected_words):
    try:
        # Dịch sang tiếng Việt
        vn_translation = translator.translate(word, dest='vi').text
    except Exception as e:
        vn_translation = word # Giữ nguyên nếu mạng lỗi
        
    try:
        # Lấy phiên âm IPA
        ipa = f"/{p.convert(word)}/"
    except:
        ipa = f"/{word}/"
        
    topic = "General Vocabulary"
    
    entry = {
        "id": start_id + i,
        "topic": topic,
        "word": word,
        "ipa": ipa,
        "vietnamese": vn_translation.lower(),
        "age_group": "6-10",
        "example": f"Look at the {word}.",
        "question1": f"What is a {word}?",
        "question2": f"Can you say {word} aloud?",
        "question3": f"Have you ever seen a {word}?",
        "question4": f"Do you like {word}?",
        "question5": f"Can you use {word} in a short sentence?",
        "image_search_url_1": f"https://commons.wikimedia.org/w/index.php?search={word}+illustration",
        "image_search_url_2": f"https://openverse.org/search/image?q={word}",
        "image_search_url_3": f"https://search.creativecommons.org/search?q={word}",
        "image_urls": [
            f"https://source.unsplash.com/400x400/?{word}",
            f"https://source.unsplash.com/400x400/?{word},cartoon",
            f"https://source.unsplash.com/400x400/?{word},cute"
        ],
        "question1_vi": f"{vn_translation.capitalize()} là gì thế nhỉ??",
        "question2_vi": f"Bé thử đọc to từ {word} xem nào??",
        "question3_vi": f"Con đã bao giờ thấy {vn_translation} chưa??",
        "question4_vi": f"Con có thích {vn_translation} không??",
        "question5_vi": f"Thử đặt một câu ngắn với từ {word} nhé??"
    }
    
    new_entries.append(entry)
    
    # In tiến độ cho dễ theo dõi
    if (i + 1) % 100 == 0:
        print(f"Đã tạo {i + 1}/{words_needed} từ...")

# 5. Gộp và xuất file mới
all_data = current_data + new_entries
output_filename = 'kids_english_dataset_10000_FINAL.json'

with open(output_filename, 'w', encoding='utf-8') as f:
    json.dump(all_data, f, ensure_ascii=False, indent=2)

print(f"Hoàn tất! File {output_filename} đã sẵn sàng với {len(all_data)} từ vựng.")
```
